## Analyse exploratoire des données

Commençons par lire le jeu de données SpaceX dans un dataframe Pandas et afficher son résumé

```python
from js import fetch
import io

URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
resp = await fetch(URL)
dataset_part_2_csv = io.BytesIO((await resp.arrayBuffer()).to_py())
df=pd.read_csv(dataset_part_2_csv)
df.head(5)
```

Voyons d'abord comment les variables `FlightNumber` (numéro de vol, qui indique la succession des tentatives de lancement) et `PayloadMass` (masse de la charge utile) influencent l'issue du lancement.

Nous pouvons tracer `FlightNumber` en fonction de `PayloadMass` et superposer le résultat du lancement. On constate que plus le numéro de vol augmente, plus le premier étage a de chances d'atterrir avec succès. La masse de la charge utile semble aussi jouer un rôle ; même avec des charges plus lourdes, le premier étage revient souvent avec succès.

```python
sns.catplot(y="PayloadMass", x="FlightNumber", hue="Class", data=df, aspect = 5)
plt.xlabel("Numéro de vol",fontsize=20)
plt.ylabel("Masse de la charge utile (kg)",fontsize=20)
plt.show()
```

Examinons maintenant chaque site de lancement pour visualiser le détail de ses lancements.

### TÂCHE 1 : visualiser la relation entre le numéro de vol et le site de lancement

Utilisez la fonction <code>catplot</code> pour tracer <code>FlightNumber</code> en fonction de <code>LaunchSite</code> : fixez le paramètre <code>x</code> à <code>FlightNumber</code>, le paramètre <code>y</code> à <code>LaunchSite</code> et le paramètre <code>hue</code> à <code>'Class'</code>.

```python
# Nuage de points : numéro de vol en x, site de lancement en y, couleur selon la classe
sns.catplot(y="LaunchSite", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Numéro de vol", fontsize=20)
plt.ylabel("Site de lancement", fontsize=20)
plt.show()
```

Essayez maintenant d'expliquer les tendances observées sur le nuage de points Numéro de vol / Site de lancement.

> **Observation (à confirmer sur votre graphique) :** sur chaque site, la proportion de succès (Class = 1) augmente avec le numéro de vol : les premiers lancements sont surtout des échecs, les plus récents surtout des réussites. Les premiers vols partent majoritairement de CCAFS SLC 40, tandis que VAFB SLC 4E et KSC LC 39A n'apparaissent que plus tard dans la série.

### TÂCHE 2 : visualiser la relation entre la masse de la charge utile et le site de lancement

Nous voulons aussi observer s'il existe une relation entre les sites de lancement et la masse des charges utiles.

```python
# Nuage de points : masse de la charge utile (kg) en x, site de lancement en y, couleur selon la classe
sns.catplot(y="LaunchSite", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Masse de la charge utile (kg)", fontsize=20)
plt.ylabel("Site de lancement", fontsize=20)
plt.show()
```

Si vous observez le nuage de points Masse de la charge utile / Site de lancement, vous constaterez que pour le site VAFB-SLC, aucune fusée n'a été lancée avec une charge utile lourde (supérieure à 10 000 kg).

### TÂCHE 3 : visualiser le taux de succès de chaque type d'orbite

Nous voulons ensuite vérifier visuellement s'il existe une relation entre le taux de succès et le type d'orbite.

Créons un `diagramme en barres` du taux de succès de chaque orbite

```python
# INDICE : groupby sur la colonne Orbit, puis moyenne de la colonne Class
success_by_orbit = df.groupby('Orbit')['Class'].mean().reset_index()

plt.figure(figsize=(10, 5))
sns.barplot(x='Orbit', y='Class', data=success_by_orbit)
plt.xlabel("Orbite", fontsize=14)
plt.ylabel("Taux de succès", fontsize=14)
plt.title("Taux de succès par type d'orbite")
plt.show()
```

Analysez le diagramme en barres pour identifier les orbites qui ont les taux de succès les plus élevés.

> **Observation (à confirmer sur votre graphique) :** les orbites ES-L1, GEO, HEO et SSO affichent un taux de succès de 100 %, alors que l'orbite SO est à 0 %. Attention : plusieurs de ces orbites ne comptent qu'un ou deux lancements, ce qui rend leur taux peu fiable.

### TÂCHE 4 : visualiser la relation entre le numéro de vol et le type d'orbite

Pour chaque orbite, nous voulons voir s'il existe une relation entre le numéro de vol et le type d'orbite.

```python
# Nuage de points : numéro de vol en x, orbite en y, couleur selon la classe
sns.catplot(y="Orbit", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Numéro de vol", fontsize=20)
plt.ylabel("Orbite", fontsize=20)
plt.show()
```

On observe que, pour l'orbite LEO, le succès semble lié au nombre de vols. À l'inverse, pour l'orbite GTO, il ne semble y avoir aucune relation entre le numéro de vol et le succès.

### TÂCHE 5 : visualiser la relation entre la masse de la charge utile et le type d'orbite

De la même façon, nous pouvons tracer les nuages de points Masse de la charge utile / Orbite pour révéler la relation entre la masse de la charge utile et le type d'orbite.

```python
# Nuage de points : masse de la charge utile en x, orbite en y, couleur selon la classe
sns.catplot(y="Orbit", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Masse de la charge utile (kg)", fontsize=20)
plt.ylabel("Orbite", fontsize=20)
plt.show()
```

Avec des charges utiles lourdes, le taux d'atterrissages réussis est plus élevé pour les orbites polaire, LEO et ISS.

En revanche, pour GTO, il est difficile de distinguer les atterrissages réussis des ratés, car les deux issues sont présentes.

### TÂCHE 6 : visualiser la tendance annuelle du taux de succès des lancements

Vous pouvez tracer un graphique en courbe avec l'année (<code>Year</code>) en abscisse et le taux de succès moyen en ordonnée, afin d'obtenir la tendance moyenne de réussite des lancements.

La fonction suivante permet d'extraire l'année à partir de la date :

```python
# Fonction qui extrait l'année à partir de la date
# (à n'exécuter qu'une fois : relancer la cellule fait échouer l'affectation, car 'Date' ne contient plus que des années)
year=[]
def Extract_year():
    for i in df["Date"]:
        year.append(i.split("-")[0])
    return year
Extract_year()
df['Date'] = year
df.head()
    
```

```python
# Courbe : année extraite en x, taux de succès moyen en y
yearly_success = df.groupby('Date')['Class'].mean().reset_index()
yearly_success['Date'] = yearly_success['Date'].astype(int)

plt.figure(figsize=(10, 5))
sns.lineplot(data=yearly_success, x='Date', y='Class', marker='o')
plt.xlabel("Année", fontsize=14)
plt.ylabel("Taux de succès moyen", fontsize=14)
plt.title("Tendance annuelle du taux de succès des lancements")
plt.show()
```

Vous pouvez observer que le taux de succès n'a cessé d'augmenter de 2013 jusqu'à 2020.

## Feature engineering

À ce stade, vous devriez avoir une première idée de l'effet de chaque variable importante sur le taux de succès. Nous allons sélectionner les variables (features) qui serviront à la prédiction du succès dans un prochain module.

```python
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 'GridFins', 'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount', 'Serial']]
features.head()
```

### TÂCHE 7 : créer des variables indicatrices (dummies) pour les colonnes catégorielles

Utilisez la fonction <code>get_dummies</code> et le dataframe <code>features</code> pour appliquer un OneHotEncoder aux colonnes <code>Orbit</code>, <code>LaunchSite</code>, <code>LandingPad</code> et <code>Serial</code>. Affectez le résultat à la variable <code>features_one_hot</code> et affichez-le avec la méthode head. Le dataframe obtenu doit contenir toutes les variables, y compris celles encodées.

```python
# INDICE : utiliser get_dummies() sur les colonnes catégorielles
features_one_hot = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad', 'Serial'])
features_one_hot.head()
```

### TÂCHE 8 : convertir toutes les colonnes numériques en `float64`

Maintenant que le dataframe <code>features_one_hot</code> ne contient que des nombres, convertissez-le entièrement au type <code>float64</code>.

```python
# INDICE : utiliser la méthode astype
features_one_hot = features_one_hot.astype('float64')
features_one_hot.head()
```
